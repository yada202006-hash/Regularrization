# Elastic net regression
Elastic Net regression is a regularization technique that combines both L1 (Lasso) and L2 (Ridge) penalties, making it especially useful when dealing with correlated features or high-dimensional datasets. It balances feature selection (from Lasso) with coefficient shrinkage (from Ridge) to improve model stability and generalization.

This is the combination of both Ridge and Lasso it decides how much percent we apply Ridge and lasso


**Use Elastic Net when you suspect multicollinearity (features highly correlated) and want both shrinkage + feature selection.**

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
x,y=load_diabetes(return_X_y=True)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
reg=ElasticNet(alpha=0.005,l1_ratio=0.9)
reg.fit(x_train,y_train)
y_pred=reg.predict(x_test)
print("r2 score",r2_score(y_test,y_pred))
```

**alpha=λ₁ + λ₂**

**l1_ratio= λ₁ / (λ₁ + λ₂)**
