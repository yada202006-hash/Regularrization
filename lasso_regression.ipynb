Lasso Regression is a type of linear regression that uses L1 regularization. Instead of just minimizing the sum of squared errors, it adds a penalty proportional to the absolute value of the coefficients.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
```

```python
x,y=make_regression(n_samples=100,n_features=1,n_informative=1,n_targets=1,noise=20,random_state=13)
plt.scatter(x,y)

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)

from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
print(lr.coef_)
print(lr.intercept_)
```

```python
alphas=[0,1,5,10,30]
plt.figure(figsize=(12,6))
plt.scatter(x,y)
for i in alphas:
    L=Lasso(alpha=i)
    L.fit(x_train,y_train)
    plt.plot(x_test,L.predict(x_test),label='alpha={}'.format(i))
plt.legend()
plt.show()
```

here, the line alpha=0 shows normal regression line when we increase alpha it is going to best fit line after incresing alpha to 30 leads to underfitting

Lasso also do feature selection it helps to reduce less feture columns alpha becomes equal to 0

# whay  lasso creates sparsity
Lasso creates sparsity because its L1 penalty includes a term with λ (lambda) in the numerator, which directly scales the strength of shrinkage. The absolute value penalty makes the optimization path hit sharp corners, and when λ is large enough, small coefficients are forced exactly to zero. In simple words: λ controls how strongly Lasso pushes weights down, and the L1 geometry ensures many of them drop out completely, giving sparsity.
