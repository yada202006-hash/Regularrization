# Bias
In the context of machine learning and statistics, bias refers to the systematic error that a model introduces because it cannot capture
the true relationship in the data. It represents the diff between the expected prediction of our model and the correct value which we are 
trying to predict. High bias leads underfiiting, where the model not fit the training data well

Bias(f'(x)) = E[f'(x)] - f(x)

f'(x)-sample data
f(x)-population data

# Bias variance decomposition
Bias–variance decomposition explains how prediction error can be split into three parts: bias² (systematic error), variance (sensitivity to training data), and irreducible noise. It’s the foundation of the bias–variance trade‑off: simple models have high bias but low variance, while complex models have low bias but high variance.
